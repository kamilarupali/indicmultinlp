# Script and language identification baseline
This notebook measures a transparent Unicode-script heuristic against a labeled test CSV. Script detection is not equivalent to language identification (e.g. Devanagari is shared by multiple languages).

```python
from pathlib import Path
import sys, pandas as pd
from sklearn.metrics import classification_report, confusion_matrix
sys.path.insert(0, str(Path('../backend').resolve()))
from app.language.language_detector import detect_language
from app.language.script_detector import detect_script
CSV_PATH = Path('../data/test/dataset.csv')  # set to your labeled test CSV
df = pd.read_csv(CSV_PATH).dropna(subset=['text', 'language'])
df['script_prediction'] = df['text'].astype(str).map(lambda text: detect_script(text)['primary'])
df['language_prediction'] = df['text'].astype(str).map(lambda text: detect_language(text)['language'])
print('Script accuracy/report (requires script labels):')
# If your gold labels identify scripts, compare them to script_prediction.
display(df[['text', 'language', 'language_prediction', 'script_prediction']].head(20))
```

## Evaluate only against matching labels
Create a `script` gold-label column for script metrics, or use `language` for language metrics. Report per-class and macro scores; inspect confusions rather than treating script guesses as language truth.

```python
if 'script' in df.columns:
    print(classification_report(df['script'].astype(str), df['script_prediction'], zero_division=0))
    print(confusion_matrix(df['script'].astype(str), df['script_prediction']))
if 'language' in df.columns:
    print('Heuristic language-label report (interpret cautiously):')
    print(classification_report(df['language'].astype(str), df['language_prediction'], zero_division=0))
```
