# N-gram language model counts
Build unigram, bigram, and trigram frequency tables from a training text column. These are count models, not supervised sentiment or intent classifiers.

```python
from collections import Counter
from pathlib import Path
import pandas as pd
CSV_PATH = Path('../data/train/dataset.csv')  # set to your training CSV
df = pd.read_csv(CSV_PATH).dropna(subset=['text'])
texts = df['text'].astype(str).tolist()
def tokens(text): return text.casefold().split()
unigrams = Counter(token for text in texts for token in tokens(text))
bigrams = Counter(pair for text in texts for pair in zip(tokens(text), tokens(text)[1:]))
trigrams = Counter(tuple(tokens(text)[i:i+3]) for text in texts for i in range(max(0, len(tokens(text))-2)))
print('vocabulary', len(unigrams), 'bigram types', len(bigrams), 'trigram types', len(trigrams))
print('common unigrams', unigrams.most_common(20))
print('common bigrams', bigrams.most_common(20))
```

```python
# Laplace-smoothed next-token probability; uses an explicit unknown-token bucket.
vocabulary_size = len(unigrams) + 1
def bigram_probability(previous, current):
    denominator = unigrams[previous] + vocabulary_size
    return (bigrams[(previous, current)] + 1) / denominator
if unigrams:
    most_common = unigrams.most_common(1)[0][0]
    print('P(common token | common previous token):', bigram_probability(most_common, most_common))
```
